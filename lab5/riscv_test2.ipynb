## Load the FPGA Design

```python
from pynq import Overlay

# change to the path where you save the bistream
overlay = Overlay('./riscv2.bit')

comm_ip = overlay.comm_0
```

```python
for name, reg in overlay.ip_dict['comm_0']['registers'].items():
    print(f"{name:10s} offset=0x{reg['address_offset']:02X}  access={reg['access']}")
```

## Dispatch Instructions

```python
# Final digit = 7
INST1 = 0x00400593 # addi x11, x0, 4 -- x11 == 4
INST2 = 0x00358513 # addi x10, x11, 3 -- x10 == 4+3=7

instructions = [INST1, INST2]

print("x10 before:", comm_ip.read(0x30))

for inst in instructions:
    comm_ip.write(0x10, inst)
    print(f"wrote 0x{inst:08X} -> x10 = {comm_ip.read(0x30)}")

out2 = comm_ip.read(0x30)
print(f"wrote 0x{INST:08X} (addi x10, x0, 7) -> x10 = {out2}")
print("expected 7 ->", "PASS" if out2 == 7 else "FAIL")
```
